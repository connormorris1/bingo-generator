```python
from create_bingo_boards import *
from pathlib import Path

image_path = "" #path to folder containing clipart images
num_boards = 1 #number of boards to generate
rows = 4
cols = 5
output_directory = Path("boards/")
target_size = 90
random_seed = None

imgs = read_imgs(image_path, target_size)

output_directory.mkdir(parents=True, exist_ok=True)

for i in range(num_boards):
    board = create_board(imgs, rows=rows, cols=cols, random_seed=random_seed)
    output_path = output_directory / f"board_{i+1}.png"
    Image.fromarray(board).save(output_path)
    print(f"Saved board {i+1} to {output_path}")
```
